```python
import torch
import torchvision
import torchvision.transforms as transforms
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
train_data = torchvision.datasets.CIFAR100(root="../data", train=True, download=True, transform=transforms.ToTensor())
test_data = torchvision.datasets.CIFAR100(root="../data", train=False, download=True, transform=transforms.ToTensor())
```

```python
classmap = sorted(list(train_data.class_to_idx.items()), key=lambda item: item[1])
```

```python
fig, axes = plt.subplots(10, 10, figsize=(12, 16))
axes = axes.ravel()

features = train_data.data
targets = np.array(train_data.targets, dtype=np.long)

for cls, id in classmap:
    idx = np.where(targets == id)[0][0]
    img = features[idx]
    axes[id].set_title(f"{cls}")
    axes[id].imshow(img)
    axes[id].axis(False)

plt.tight_layout()
plt.show()
```
